```python
import gradio as gr
import cv2
import numpy as np
```

```python

def load_image(path):
    image = cv2.imread(path)
    return image
```

```python

def convert_to_grayscale(image):
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    return gray, gray  # keep second gray for state
```

```python

def apply_adaptive_threshold(image):
    thresholded = cv2.adaptiveThreshold(image, 255,
                                        cv2.ADAPTIVE_THRESH_GAUSSIAN_C,
                                        cv2.THRESH_BINARY, 11, 2)
    return thresholded, thresholded
```

```python

def reduce_noise(image):
    denoised = cv2.medianBlur(image, 5)
    return denoised, denoised
```

```python

def morphological_operations(image):
    kernel = np.ones((1, 1), np.uint8)
    morphed = cv2.dilate(cv2.erode(image, kernel, iterations=1), kernel, iterations=1)
    return morphed
```

```python

with gr.Blocks() as demo:
    gr.Markdown("## Invoice Preprocessing Pipeline (Single Image)")

    with gr.Row():
        with gr.Column():
            upload = gr.Image(type="filepath", label="Upload Invoice")
            process_button = gr.Button("Process")
        with gr.Column():
            grayscale = gr.Image(label="1. Grayscale")
            shadow_removed = gr.Image(label="2. Thresholded")
            denoised = gr.Image(label="3. Noise Reduced")
            binary = gr.Image(label="4. Morphological")

    gray_state = gr.State()
    thresh_state = gr.State()
    noise_state = gr.State()

    process_button.click(fn=load_image, inputs=upload, outputs=gray_state) \
        .then(fn=convert_to_grayscale, inputs=gray_state, outputs=[grayscale, gray_state]) \
        .then(fn=apply_adaptive_threshold, inputs=gray_state, outputs=[shadow_removed, thresh_state]) \
        .then(fn=reduce_noise, inputs=thresh_state, outputs=[denoised, noise_state]) \
        .then(fn=morphological_operations, inputs=noise_state, outputs=binary)

demo.launch()
```
